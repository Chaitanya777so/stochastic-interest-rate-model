```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/fedfunds.csv")

df["observation_date"] = pd.to_datetime(df["observation_date"])

df.set_index("observation_date", inplace=True)

df.head()
```

```python
df.info()
```

```python
df["lag_1"] = df["FEDFUNDS"].shift(1)
```

```python
df["lag_2"] = df["FEDFUNDS"].shift(2)
```

```python
df["lag_3"] = df["FEDFUNDS"].shift(3)
```

```python
df.head(10)
```

```python
df["rolling_mean_3"] = (
    df["FEDFUNDS"]
    .rolling(window=3)
    .mean()
)
```

```python
df["rolling_mean_12"] = (
    df["FEDFUNDS"]
    .rolling(window=12)
    .mean()
)
```

```python
df["rate_change"] = (
    df["FEDFUNDS"]
    .diff()
)
```

```python
df["pct_change"] = (
    df["FEDFUNDS"]
    .pct_change()
)
```

```python
df["volatility_12"] = (
    df["FEDFUNDS"]
    .rolling(window=12)
    .std()
)
```

```python
df["trend"] = np.where(
    df["rate_change"] > 0,
    1,
    0
)
```

```python
df["year"] = df.index.year

df["month"] = df.index.month

df["quarter"] = df.index.quarter
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df.to_csv(
    "../data/processed/fedfunds_features.csv"
)
```
